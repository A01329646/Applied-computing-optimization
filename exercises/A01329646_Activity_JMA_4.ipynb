## Table 1, ID 6: Damped interaction potential### Function definition

### Function: $f(x,y) = \cos(x)\cos(y)e^{-\frac{x^2+y^2}{10}}$

**Problem Type:** Maximization  
**Range:** $[-10, 10]$  
**Characteristics:** Radially damped oscillations

```python
x = np.linspace(-10,10,200)
y = np.linspace(-10,10,200)

X,Y = np.meshgrid(x,y)

Z = np.cos(X)*np.cos(Y)*np.exp(-(X**2+Y**2)/10)
```

### Plot in 2D

```python
plt.figure(figsize=(6,5))
plt.contourf(X,Y,Z,50)
plt.colorbar()
plt.xlabel("x")
plt.ylabel("y")
plt.title("Potential Surface")
plt.show()
```

### Plot in 3D

```python
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

surface = ax.plot_surface(
    X, Y, Z,
    cmap='viridis',
    edgecolor='none',
    alpha=0.9
)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('f(x,y)')
ax.set_title('Surface Plot')

fig.colorbar(surface, shrink=0.5, aspect=10)

plt.show()
```

### Random search optimization

```python
import random
import math

# Define the function to maximize
def objective_function(x, y):
    return np.cos(x) * np.cos(y) * np.exp(-(x**2 + y**2) / 10)

# Initialize with a random starting point in the search space
search_range = 10  # Search in [-10, 10] for both x and y
best_x = random.uniform(-search_range, search_range)
best_y = random.uniform(-search_range, search_range)
best_value = objective_function(best_x, best_y)

print(f"Initial point: ({best_x:.4f}, {best_y:.4f})")
print(f"Initial value: {best_value:.6f}\n")

max_dividend = 200
number_of_points = 2000

# max_dividend is the number of iterations we want to perform, it also controls how fine we want to search for the solution, the larger n is, the smaller the neighborhood we search in
for n in range(1, max_dividend):
    # Generate random points around the best solution, with a neighborhood that gets smaller as n increases
    for i in range(number_of_points):
        # Search around the best solution with a neighborhood that gets smaller as n increases
        x_new = random.uniform(best_x - search_range/n, best_x + search_range/n)
        y_new = random.uniform(best_y - search_range/n, best_y + search_range/n)
        
        # Evaluate the function at the new point
        value = objective_function(x_new, y_new)
        
        # Check if we found a better solution (MAXIMUM, so we use >)
        if value > best_value:
            best_value = value
            best_x = x_new
            best_y = y_new
            print(f"Iteration {n}: Best value = {best_value:.6f} at ({best_x:.4f}, {best_y:.4f})")

print(f"\n{'='*60}")
print(f"Best maximum found: {best_value:.6f}")
print(f"At position: x = {best_x:.4f}, y = {best_y:.4f}")
print(f"{'='*60}")

# Visualize the result on the contour plot
plt.figure(figsize=(8, 6))
plt.contourf(X, Y, Z, 50, cmap='viridis')
plt.colorbar(label='f(x,y)')
plt.scatter(best_x, best_y, color='red', s=200, marker='*', 
            edgecolors='white', linewidths=2, label=f'Maximum: {best_value:.6f}')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Random Search Optimization Result')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Table 2, ID 6: Easom Function
### Function: $f(x,y) = -\cos(x)\cos(y)e^{-((x-\pi)^2+(y-\pi)^2)}$

**Problem Type:** Minimization  
**Range:** $[-10, 10]$  
**Characteristics:** Very sharp global minimum at $(\pi, \pi)$

### Function definition

```python
x2 = np.linspace(-10, 10, 200)
y2 = np.linspace(-10, 10, 200)

X2, Y2 = np.meshgrid(x2, y2)

Z2 = -np.cos(X2) * np.cos(Y2) * np.exp(-((X2 - np.pi)**2 + (Y2 - np.pi)**2))
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

surface = ax.plot_surface(
    X2, Y2, Z2,
    cmap='viridis',
    edgecolor='none',
    alpha=0.9
)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('f(x,y)')
ax.set_title('Easom Function - 3D Surface Plot')

fig.colorbar(surface, shrink=0.5, aspect=10)

plt.show()
```

### Random search optimization

```python
import random
import math

# Define the Easom function to MINIMIZE
def easom_function(x, y):
    return -np.cos(x) * np.cos(y) * np.exp(-((x - np.pi)**2 + (y - np.pi)**2))

# Initialize with a random starting point in the search space
search_range_easom = 10  # Search in [-10, 10] for both x and y
best_x_easom = random.uniform(-search_range_easom, search_range_easom)
best_y_easom = random.uniform(-search_range_easom, search_range_easom)
best_value_easom = easom_function(best_x_easom, best_y_easom)

print(f"Initial point: ({best_x_easom:.4f}, {best_y_easom:.4f})")
print(f"Initial value: {best_value_easom:.6f}\n")

max_dividend_easom = 200
number_of_points_easom = 2000

# Stochastic optimization loop
for n in range(1, max_dividend_easom):
    # Generate random points around the best solution
    for i in range(number_of_points_easom):
        # Search around the best solution with a neighborhood that gets smaller as n increases
        x_new = random.uniform(best_x_easom - search_range_easom/n, best_x_easom + search_range_easom/n)
        y_new = random.uniform(best_y_easom - search_range_easom/n, best_y_easom + search_range_easom/n)
        
        # Evaluate the function at the new point
        value = easom_function(x_new, y_new)
        
        # Check if we found a BETTER solution (MINIMUM, so we use <)
        if value < best_value_easom:
            best_value_easom = value
            best_x_easom = x_new
            best_y_easom = y_new
            print(f"Iteration {n}: Best value = {best_value_easom:.6f} at ({best_x_easom:.4f}, {best_y_easom:.4f})")

print(f"\n{'='*60}")
print(f"Best minimum found: {best_value_easom:.6f}")
print(f"At position: x = {best_x_easom:.4f}, y = {best_y_easom:.4f}")
print(f"{'='*60}")

# Visualize the result on the contour plot
plt.figure(figsize=(8, 6))
plt.contourf(X2, Y2, Z2, 50, cmap='viridis')
plt.colorbar(label='f(x,y)')
plt.scatter(best_x_easom, best_y_easom, color='red', s=200, marker='*', 
            edgecolors='white', linewidths=2, label=f'Minimum: {best_value_easom:.6f}')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Easom Function - Random Search Optimization Result')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
